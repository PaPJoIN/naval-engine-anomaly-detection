A low lubrication oil temperature may indicate it is not reaching its optimal operating temperature, potentially causing inadequate lubrication.
- **Coolant temperature:** High coolant temperature signals overheating, which various issues, including a failed thermostat, coolant leak, or insufficient coolant flow can cause. A low coolant temperature could suggest the engine is not reaching its optimal operating temperature, affecting performance and efficiency.

Issues with engines could lead to engine malfunctions, potential safety hazards, and downtime (e.g. delayed deliveries), resulting in the breakdown of a ship’s overall functionality, consequently impacting the business, such as affecting revenue via failure to deliver goods. # Anomaly detection: Statistical methods

```python
# Created a custom function to apply the IQR method for identifying outliers
def find_outliers(df):
  """This function loops over the columns to calculate the IQR, lower/upper ranges,
    and creates the new columns that contain binary values
    which indicate the observations that lie beyond the Upper and lower IQR ranges"""
  for col in df.columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = sci.iqr(df[col])
    lower_limit = q1-1.5*iqr
    upper_limit = q3+1.5*iqr
    # Created & populated the new columns in the dataframe
    df[f"{col}_Outlier"] = ((df[col] > upper_limit)|(df[col] < lower_limit)).astype(int)
  return df

find_outliers(engines)
```

```python
# Selected only the columns which have 1 or more outliers
outlier_columns = ['Engine rpm_Outlier', 'Lub oil pressure_Outlier', 'Fuel pressure_Outlier', 'Coolant pressure_Outlier', 'lub oil temp_Outlier', 'Coolant temp_Outlier']
outlier_columns = engines.filter(like="_Outlier").columns
engines.loc[:,['Engine rpm_Outlier', 'Lub oil pressure_Outlier', 'Fuel pressure_Outlier', 'Coolant pressure_Outlier', 'lub oil temp_Outlier', 'Coolant temp_Outlier']]

#Alternative: select only columns having outlier in the name
#regex='_Outlier$': This regular expression pattern matches any column name that ends with '_Outlier'. The $ symbol ensures it matches only at the end of the column name.
engines.filter(like="_Outlier")
engines.filter(regex="_Outlier$")

#Alternative: select only rows having outliers in the column name AND 1 or more values in the row
engines.filter(like="_Outlier").sum(axis=1) > 0
engines.filter(like="_Outlier").sum(axis=1) > 1
```

```python
#Created a column that sums the values of the outlier columns for each row
outlier_columns = engines.filter(like="_Outlier").columns
engines["Outlier_Count"]=engines[outlier_columns].sum(axis=1)

display(engines)
```

```python
# Displayed how many rows have an outlier count value of 0,1,2 and 3
engines["Outlier_Count"].value_counts()
```

```python
# Created a new dataframe containing only observations with outlier count equal to 2 or above
IQR_outliers = engines[engines["Outlier_Count"] > 1]
display(IQR_outliers)
```

```python
# Calculated the Percentage of the samples with feature outlier counts greaters than 1
outlier_percentage = (len(IQR_outliers) / len(engines)) * 100
print(f"Percentage of outliers: {outlier_percentage:.2f}%") # {variable:.2f} -> speficies to round the value of the variable to 2 decimal points
```

Initially I began calculating the IQR and ranges for each feature manually, but this proved to be a somewhat inefficient and time consuming. Therefore, I decided to optimise and simplify the process by creating a custom function that loops over the data and created the necessary columns. While there are more efficient ways to conduct anomaly detection, this function has the benefit of transferability to other similar cases.

After creating the custom function for the IQR anomaly detection and the corresponding columns in the datadframe, I isolated the rows that have more that 1 ouliers, which returned a dataframe of **422 rows**. 411 of those were entries with 2 features containing outliers values and 11 rows contained 3 outlier values.

# Anomaly detection: Machine Learning methods

### One-class SVM

```python
# Imported the necessary modules
from sklearn.preprocessing import StandardScaler
from sklearn.svm import OneClassSVM
from sklearn.decomposition import PCA

engines_ML = pd.read_csv('https://raw.githubusercontent.com/fourthrevlxd/cam_dsb/main/engine.csv')
```

```python
# Scaled the Data and stored it into the new dataframe
scaler = StandardScaler()
engines_svm_scaled = pd.DataFrame(scaler.fit_transform(engines_ML), columns=engines_ML.columns)
#Added a column that will store the predicted values of the SVM
engines_svm_scaled["Anomaly"] = 0
engines_svm_scaled.head()
```

```python
# Defined the X variables
X = engines_svm_scaled.iloc[:, :-1]
```

Trained and applied the model using
- Gamma = 0.5
- Nu = 0.05

```python
# Defined & trained the model with default parameters
model_svm = OneClassSVM(kernel="rbf", gamma=0.5, nu=0.05)
model_svm.fit(X)

# Applied the model to the Dataset
engines_svm_scaled["Anomaly"] = model_svm.predict(X)
display(engines_svm_scaled)

# Performed a count on the normal vs anomaly values
print("\n","Number of anomalies (-1):")
engines_svm_scaled["Anomaly"].value_counts()
```

```python
# Calculated the Anomaly percentage
percentage = (len(engines_svm_scaled[engines_svm_scaled["Anomaly"] == -1]) / len(engines_ML)) * 100
print(f"Percentage of Anomalies: {percentage:.2f}%")
```

Re-trained and applied the model using
- Gamma = 0.3
- Nu = 0.05

```python
# Re-trained the model using lower values for Gamma
model_svm = OneClassSVM(kernel="rbf", gamma=0.3, nu=0.05)
model_svm.fit(X)

# Applied the model to the Dataset
engines_svm_scaled["Anomaly"] = model_svm.predict(X)

# Performed a count on the normal vs anomaly values
print("\n","Number of anomalies (-1):")
engines_svm_scaled["Anomaly"].value_counts()
```

```python
# Calculated the Anomaly percentage
percentage = (len(engines_svm_scaled[engines_svm_scaled["Anomaly"] == -1]) / len(engines_ML)) * 100
print(f"Percentage of Anomalies: {percentage:.2f}%")
```

Re-trained and applied the model using
- Gamma = 0.2
- Nu = 0.03

```python
# Re-trained the model using lower values for Gamma & Nu
model_svm = OneClassSVM(kernel="rbf", gamma=0.2, nu=0.03)
model_svm.fit(X)

# Applied the model to the Dataset
engines_svm_scaled["Anomaly"] = model_svm.predict(X)

# Performed a count on the normal vs anomaly values
print("\n","Number of anomalies (-1):")
engines_svm_scaled["Anomaly"].value_counts()
```

```python
# Calculated the Anomaly percentage
percentage = (len(engines_svm_scaled[engines_svm_scaled["Anomaly"] == -1]) / len(engines_ML)) * 100
print(f"Percentage of Anomalies: {percentage:.2f}%")
```

By applying the model using Gamma = 0.3 and Nu = 0.05 I received an amonaly percentage of **3.02** which would imply it is the best fit considering the bounds between 1-5%

```python
#Rerformed PCA on the engine features
pca = PCA(n_components=2)
engines_pca = pca.fit_transform(X)
print(engines_pca.shape)
print(engines_pca)
```

```python
# Created a scatter plot Principle components with the colored anomaly predictions
plt.figure(figsize=(8, 6))
sns.scatterplot(x=engines_pca[:, 0], y=engines_pca[:, 1], hue=engines_svm_scaled["Anomaly"], palette="viridis") # (c, cmap) & plt.legend() when using plt instead!
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("One-Class SVM Anomaly detection")
handles, labels_  =  plt.gca().get_legend_handles_labels()
plt.legend(handles, ['Anomaly','Normal'], title="Observation")

plt.show()
```

### Isolation Forest

```python
# Imported the necessary module
from sklearn.ensemble import IsolationForest
```

```python
# Created a new Dataframe with an additional column to store the prediction results
engines_if = engines_ML
engines_if["Anomaly_IF"] = 0
engines_if.head()
```

```python
# Defined the unscaled data for the Isolation forest model
X1 = engines_if.iloc[:,:-1]
display(X1)
```

Trained and applied the model using:
- Contamination = 0.05
- n_estimators = 100

```python
#Defines the IF model
model_forest = IsolationForest(n_estimators=100,contamination=0.05,random_state=42) #Provides 5% anomalies

# Trained the model
model_forest.fit(X1)

# Calculated the anomaly score of datapoints
model_forest.decision_function(X1)
```

```python
# Applied model to perform predictions and
engines_if["Anomaly_IF"] = model_forest.predict(X1)
display(engines_if)
# Performed a count on the normal vs anomaly values
engines_if["Anomaly_IF"].value_counts()
```

```python
# Calculated the Anomaly percentage
percentage_if = (len(engines_if[engines_if["Anomaly_IF"] == -1]) / len(engines_ML)) * 100
print(f"Percentage of Anomalies: {percentage_if:.2f}%")
```
